# **TITANIC ANALYSIS**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import plotly.express as px
```

```python
ds=pd.read_csv(r"/content/titanic.csv")
print(ds)
```

```python
print(ds.head())
```

```python
ds["Fare"]=ds["Fare"].fillna(ds["Fare"].mean())
print(ds)
```

```python
ds["Age"]=ds["Age"].fillna(ds["Age"].mean())
print(ds)
```

```python
ds["PassengerId"]=ds["PassengerId"].fillna(ds["PassengerId"].mean())
print(ds)
```

```python
ds["SibSp"]=ds["SibSp"].fillna(ds["SibSp"].mean())
print(ds)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x="Survived", data=ds)
plt.title("Survival Counts")
plt.show()
```

```python
sns.barplot(x="Sex", y="Survived", data=ds)
plt.title("Survival by Gender")
plt.show()
```

```python
plt.hist(ds["Age"].dropna(), bins=20,color='g')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.scatterplot(x="Pclass", y="Fare", data=ds)
plt.title("Compare Fare by Passenger Class")
plt.show()
```

```python
sns.boxplot(x="Age", y="Fare", data=ds)
plt.title("Compare Fare by Age")
plt.show()
```

```python
sns.boxplot(y="Fare", data=ds)
plt.title("Fare Outliers")
plt.show()
```

```python
ds.describe()
```

```python
numeric_ds = ds.select_dtypes(include=["number"])
plt.figure(figsize=(8,6))
sns.heatmap(numeric_ds.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
